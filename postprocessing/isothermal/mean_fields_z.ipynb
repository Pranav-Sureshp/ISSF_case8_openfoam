```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
simulation_file = './y0_xm03_umean.csv'

nekrs_data = '/home/pranav/OpenFOAM_cases/research/ISSF_isotherm/NekRS_results/lines/lineZ_y0_xm03_Uav_Code.csv'
```

```python
#nekrs_scales #
u_ref_nekrs = 0.03 ;
p_ref_nekrs = 8.854209 ;
L_ref_nekrs = 0.023 ;
```

```python
# 1. Read the CSV file (assumes 1st line is the header)
df = pd.read_csv(simulation_file); nekrs = pd.read_csv(nekrs_data) ;
df.fillna(0, inplace=True)
# 2. Define the columns you want to plot
# Replace these with your actual column names from the CSV header
x_col = 'Points_2'
y_col = 'UMean_0'

# 3. Create the plot with big font sizes
plt.figure(figsize=(10, 6))

# Use the plot function from matplotlib
plt.plot(df[x_col], df[y_col],'r-*',markevery=10,label=f'OpenFOAM', linewidth=2)
plt.plot(nekrs[x_col], nekrs['Velocity_0'] * (u_ref_nekrs),'kd',markevery=10,label=f'NEKRS ', linewidth=2)

# 4. Set labels and title with big fontsize
plt.xlabel(r'z', fontsize=18) ; plt.ylabel(r'$\overline{U}$ (m/s)', fontsize=18)
plt.title(r'x=-0.3m',fontsize=18)
# 5. Add a legend with big fontsize
plt.legend(fontsize=16, loc='best')

# 6. Increase tick label sizes
plt.xticks(fontsize=14) ; plt.yticks(fontsize=14)

plt.grid(True)
plt.tight_layout()
plt.savefig('./Umean_z_y0_xm03.png', dpi=300,bbox_inches='tight')
```
